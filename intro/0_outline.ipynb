# 整体流程
## 1. 加载环境变量
读取项目中的 `.env` 文件，把里面的配置加载到环境变量中，方便代码使用 API Key 等信息

````python
from dotenv import load_dotenv

load_dotenv()
````

## 2. 定义工具

````python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
	"""
	Get the weather in a given location.
	Args:
		location: city name or coordinates
	"""
	return f"The weather in {location} is sunny."
````

## 3. 创建 Agent

````python
from langchain.agents import create_agent

agent = create_agent(
	"gpt-4o-mini",
    tools=[get_weather]
)
````

## 4. 发起调用
```
User                   Agent                                             Model
|                        |    将用户问题 message,                           |
|- What is the weather ->|--- 工具信息 tools {                              |
|  in New York?          |   |  name: get_weather                          |
|                        |   |  description: Get the weather               |
|                        |   |               for a given location.         |
|                        |   |  parameter: location                        |
|                        |<-- } 组装为请求参数                               |
|                        |                                                 |
|                        |-------------------- 调用模型 ------------------->|
|                        |                                                 |---
|                        |                                                 |   | 判断是否调用工具
|                        |                                                 |   | 及调用哪个
|                        |                                                 |<--
|                        |<-------------- 调用工具 get_weather -------------|
|                        |---                                              |
|                        |   | 执行工具                                     |
|                        |<--                                              |
|                        |- 返回工具结果 The weather in New York is sunny. ->|
|                        |                                                 |---
|                        |                                                 |   | 分析结果
|                        |                                                 |   | 生成自然语言回答
|                        |                                                 |<--
|                        |      返回答案 The weather in New York is sunny.  |
|   The weather in New   |<---- What else do you want to know? ------------|
|   York is sunny. What  |                                                 |
|<- else do you want to -|                                                 |
|   know?                |                                                 |

```

````python
response = agent.invoke({
	"messages": [
		{"role": "user", "content": "What is the weather in New York?"}
	]
})
````
