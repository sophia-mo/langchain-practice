### 1.2 使用 LangChain 不支持的国内模型

````python
import os
from langchain.chat_models import init_chat_model

base_url = os.getenv("DASHSCOPE_BASE URL")
api_key = os.getenv("DASHSCOPE_API_KEY")
model = init_chat_model(
	model="qwen-max",			# 模型名称, 可自定义, 这里用的是阿里的 qwen-max
	model_provider="openai",	# 阿里云百炼兼容 openai
	base_url=base_url,		# 在阿里云百炼中查询
	api_key=api_key,
	temparature=1.5,			# 控制生成文本的随机性, 值越小越确定, 值越大越随机
	top_p=0.9					# 控制生成文本的多样性, 值越小越多样, 值越大越确定
	# max_tokens: 控制生成文本的最大长度
	# timeout: 控制生成文本的超时时间
	# max_retries: 控制生成文本的最大重试次数
)

print(type(model))
````

## 2. 模型调用
### 2.1 Invoke 阻塞式

````python
response = model.invoke("What is the weather in New York?")
print(response.content)
````

### 2.2 Stream 流式

````python
stream = model.stream("What is the weather in New York?")
for chunk in stream:
	print(chunk.content, end="", flush=True)
````

## 3. 在 Agent 中使用模型
### 3.1 创建 Agent

````python
from langchain.agents import create_agent

# 3.1.1 使用初始化好的 model
agent = create_agent(model=model)

# 3.1.2 指定 model 名称, 由 LangChain 自动初始化
agent = create_agent(model="gpt-4o-mini")
````

### 3.2 调用 Agent
#### 3.2.1 Invoke 阻塞式调用

````python
response = agent.invoke({
    "messages": [{"role": "user", "content": "Who are you?"}]
})
print(response.content)
````

#### 3.2 Stream 流式访问

````python
messages = agent.stream(
	{"messages": [{"role": "user","content":"Who are you?"}]},
	stream_mode="messages"
)
print(type(messages))
````

````python
for token, metadata in messages:
	if token.content:	# Check if there's actual content
		print(token.content, end="", flush=True)
		# 打印后不自动换行. 默认是 end="\n", 会导致每个片段单独占一行
		# flush=True 确保立即输出, 而不是缓冲后再输出
````
